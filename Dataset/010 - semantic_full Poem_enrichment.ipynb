```python
import pandas as pd

# --- Step 1: Load CSVs
csv_01_path = ".\CSV\Diwan-Hamdan-WIP - Full_poems.csv"  # poem-level CSV
csv_02_path = ".\CSV\Diwan-Hamdan-WIP - Semantic_search.csv"  # chunk-level CSV

df_full = pd.read_csv(csv_01_path)
df_chunks = pd.read_csv(csv_02_path)

print(f"[INFO] Loaded csv_01: {len(df_full)} poems")
print(f"[INFO] Loaded csv_02: {len(df_chunks)} chunk rows")

# --- Step 2: Build lookup for full poems
full_poem_map = df_full.set_index("poem_id")["Poem_line_cleaned"].to_dict()
print(f"[INFO] Created full_poem lookup for {len(full_poem_map)} poems")

# --- Step 3: Enrich csv_02 with full_poem
df_chunks["full_poem"] = df_chunks["poem_id"].map(full_poem_map)

# --- Step 4: Logging some info
unique_poems = df_chunks["poem_id"].nunique()
print(f"[INFO] Number of unique poems in csv_02: {unique_poems}")
print(f"[INFO] Total chunks in csv_02: {len(df_chunks)}")

# Show sample enrichment for first 3 chunks
print("\n[INFO] Sample enriched rows:")
print(df_chunks.head(3).to_string(index=False))

# --- Step 5: Save enriched CSV
output_path = "csv_02_enriched.csv"
df_chunks.to_csv(output_path, index=False)
print(f"\n[SUCCESS] Enriched CSV saved to: {output_path}")
```
