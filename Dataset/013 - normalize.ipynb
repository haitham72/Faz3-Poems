```python
%pip install pyarabic
import pyarabic.araby as araby

def perfect_normalize(text):
    # 1. Strip Tashkeel (vowels) and Tatweel (stretch lines)
    text = araby.strip_tashkeel(text)
    text = araby.strip_tatweel(text)
    
    # 2. Normalize Alef variants (أ, إ, آ -> ا)
    # This ensures "الأنبياء" matches "الانبياء"
    text = araby.normalize_alef(text)
    
    # 3. Normalize Teh Marbuta (ة -> ه)
    # This ensures "المحرمة" matches "المحرمه"
    text = araby.normalize_teh(text)
    
    # 4. Normalize Hamza variants (ؤ, ئ -> ء) 
    # This fixes the "ID breaking" issue you had earlier
    text = araby.normalize_hamza(text)
    
    # 5. Final clean: Remove extra whitespace/tabs
    text = " ".join(text.split())
    
    return text

# Test
test_str = u"وجديـلها ياقـلب ويش انت شـايف      لـيـل ٍ يـزيـد اللـي غـيا فـيـه غـيّـا"
print(f"Cleaned: {perfect_normalize(test_str)}")
```

```python
%pip install pyarabic
import pandas as pd
import pyarabic.araby as araby
import re

def perfect_normalize(text):
    if pd.isna(text):
        return text

    # 1. Remove Tashkeel
    text = araby.strip_tashkeel(text)

    # 2. Remove Tatweel
    text = araby.strip_tatweel(text)

    # 3. Normalize Alef variants (أ إ آ -> ا)
    text = araby.normalize_alef(text)

    # 4. Normalize Teh Marbuta (ة -> ه)
    text = araby.normalize_teh(text)

    # 5. Normalize Hamza variants
    text = araby.normalize_hamza(text)

    # 6. Remove only tabs, NOT multiple spaces
    text = text.replace("\t", " ")

    # 7. Remove accidental double spaces EXCEPT 5-space poetic gap
    # (preserve 5 consecutive spaces exactly)
    text = re.sub(r'(?<!\s) {2,4}(?!\s)', ' ', text)

    return text

```
