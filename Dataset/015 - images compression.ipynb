```python
import os
from PIL import Image

# =========================
# CONFIG
# =========================
ROOT_PATH = r"../images/edit"

FOCUS_QUALITY = 60
THUMB_QUALITY = 60
SUPPORTED_EXT = (".png", ".jpg", ".jpeg", ".webp")

# =========================
# IMAGE COMPRESS FUNCTION
# =========================
def compress_to_jpg(image_path, quality):
    try:
        with Image.open(image_path) as img:
            rgb_img = img.convert("RGB")
            new_path = os.path.splitext(image_path)[0] + ".jpg"

            rgb_img.save(
                new_path,
                "JPEG",
                quality=quality,
                optimize=True
            )

        if image_path != new_path and os.path.exists(image_path):
            os.remove(image_path)

        return True

    except Exception as e:
        print(f"Failed: {image_path} | Error: {e}")
        return False


# =========================
# PROCESS SUBFOLDERS
# =========================
def process_directory(root_path):
    for root, dirs, files in os.walk(root_path):

        # Process Focus folders
        if os.path.basename(root).lower() == "focus":
            count = 0
            for file in files:
                if file.lower().endswith(SUPPORTED_EXT):
                    if compress_to_jpg(os.path.join(root, file), FOCUS_QUALITY):
                        count += 1
            if count > 0:
                print(f"Focus: {count} images found and compressed.")

        # Process thumbnail/downscaled folders
        if root.lower().endswith(os.path.join("thumbnail", "downscaled").lower()):
            count = 0
            for file in files:
                if file.lower().endswith(SUPPORTED_EXT):
                    if compress_to_jpg(os.path.join(root, file), THUMB_QUALITY):
                        count += 1
            if count > 0:
                print(f"thumbnail/downscaled: {count} images found and compressed.")


if __name__ == "__main__":
    process_directory(ROOT_PATH)
```
